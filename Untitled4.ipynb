```python
from datetime import datetime, timezone

class ConversationManager:
    def __init__(self, client=None, model="llama3-8b-8192"):
        self.client = client
        self.model = model
        self.messages = []

    def add_message(self, role, content):
        self.messages.append({
            "role": role,
            "content": content,
            "timestamp": datetime.now(timezone.utc).isoformat()
        })

    def summarize_region_with_model(self, messages, summary_instructions, max_tokens=256):
        if not self.client:
            raise RuntimeError("No client configured for summarization.")

        combined = "\n\n".join(f"{m['role']}: {m['content']}" for m in messages)
        prompt = f"{summary_instructions}\n\n{combined}"

        response = self.client.chat.completions.create(
            model=self.model,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=max_tokens
        )
        return response.choices[0].message.content.strip()

    def periodic_summarize(self, k=3, summary_instructions="Summarize recent messages."):
        if len(self.messages) < k:
            return None
        region = self.messages[-k:]
        summary = self.summarize_region_with_model(region, summary_instructions)
        self.add_message("system", f"Summary: {summary}")
        return summary
```

```python
mgr = ConversationManager(client=client, model="llama-3.3-70b-versatile")

convos = [
    ("user", "Hello, I want to learn Python."),
    ("assistant", "Python is great for beginners. Do you want resources?"),
    ("user", "Yes, please share some tutorials."),
    ("assistant", "Here are some."),
    ("user", "Thanks, also can you show me some practice problems?")
]

for role, content in convos:
    mgr.add_message(role, content)
    if role == "user":
        s = mgr.periodic_summarize(
            k=3,
            summary_instructions="Create a concise summary of the last few messages for context."
        )
        if s:
            print("\n--- Periodic summarization happened ---\n", s, "\n--- end summary ---\n")

print("\nFinal conversation history:")
for m in mgr.messages:
    print(f"{m['role']} — {m['content']}")
```

```python
# ======================================
# 📌 Cell 4: Task 2 – JSON Schema Classification
# ======================================
schema = {
    "name": "extract_user_info",
    "description": "Extracts basic user info from a chat",
    "parameters": {
        "type": "object",
        "properties": {
            "name": {"type":"string", "description":"Full name of user"},
            "email": {"type":"string", "description":"Email address of user"},
            "phone": {"type":"string", "description":"Phone number of user"},
            "location": {"type":"string", "description":"Location or city of user"},
            "age": {"type":"string", "description":"Age of user"}
        },
        "required": ["name","email","phone","location","age"]
    }
}

def extract_info(chat_text, model="llama-3.3-70b-versatile"):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role":"system","content":"You are an information extraction assistant."},
            {"role":"user","content":chat_text}
        ],
        functions=[schema],
        function_call={"name":"extract_user_info"}
    )
    return response.choices[0].message.function_call.arguments
```

```python
# ======================================
# 📌 Cell 5: Task 2 – Demo with sample chats
# ======================================
sample_chats = [
    "Hi, my name is John Doe. I live in New York, I'm 29 years old. My email is john@example.com and my phone number is 1234567890.",
    "Hello, I'm Priya Sharma from Mumbai. I'm 22 years old. Contact me at priya@gmail.com or 9876543210.",
    "Hey, this is Carlos. I stay in Madrid. I'm 34, email is carlos@mail.com and phone is 555123456."
]

results = []
for chat in sample_chats:
    extracted = extract_info(chat)
    results.append(extracted)

import pandas as pd
df = pd.DataFrame(results)
print("\n Extracted structured info:\n")
display(df)
```
